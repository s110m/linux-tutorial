## 1. What Is Linux?

Strictly speaking, **Linux** is a *kernel* - the core program that manages a computer's hardware (CPU scheduling, memory, devices, filesystems) and exposes it to other programs through system calls. What most people call "Linux" (Ubuntu, Fedora, Debian, and so on) is really the Linux kernel bundled with a large collection of other software - shells, compilers, editors, libraries, and system utilities - much of it from the **GNU Project**. That's why some people prefer the more precise name **GNU/Linux**.

Linux is **free and open-source software**: its source code is publicly available, and anyone can read it, modify it, and redistribute it, subject to its license (the kernel is licensed under the GPLv2).


## 2. A Brief History of Linux

- **1969-1970** - **Unix** is created at Bell Labs by Ken Thompson and Dennis Ritchie. Its design - a small kernel, a hierarchical filesystem, "everything is a file", small composable tools connected by pipes - shaped essentially every operating system that followed it, including Linux.
- **1983** - Richard Stallman launches the **GNU Project**, aiming to build a complete, free Unix-like operating system ("GNU's Not Unix"). By the late 1980s GNU had produced a compiler (`gcc`), an editor (`emacs`), a shell (`bash`), and most of the core userland tools - but it still lacked a working kernel.
- **1991** - Linus Torvalds, a Finnish student, starts writing a kernel as a hobby project and announces it on Usenet. He releases it under the GPL, and it quickly gets combined with the GNU userland to form a complete, free operating system.
- **1990s** - Early distributions appear (Slackware 1993, Debian 1993, Red Hat 1994) to package the kernel and GNU tools together with an installer, making Linux practical for non-experts to install.
- **2000s-today** - Linux becomes the dominant operating system for servers and cloud infrastructure, the basis of Android (the most-used mobile OS), and runs on nearly all of the world's top supercomputers, alongside a smaller but active desktop presence.


## 3. Why Linux Matters

- **Servers and cloud** - the large majority of public cloud virtual machines and web servers run Linux, because it is free, stable, scriptable, and well suited to running headless (without a graphical desktop).
- **Embedded systems** - routers, smart TVs, cars, and IoT devices frequently run a minimal Linux kernel because it is small, portable, and free of licensing fees.
- **Mobile** - Android is built on the Linux kernel.
- **Supercomputing** - virtually all of the TOP500 supercomputers run Linux, due to its performance, configurability, and lack of licensing costs at scale.
- **Development and DevOps** - most modern software delivery (containers, CI/CD, cloud infrastructure) is built assuming a Linux environment, which is why Linux administration skills remain foundational for developers and operators alike.


## 4. The Pieces That Make Up a Linux System

A running Linux system is made of several layers stacked on top of each other:

| Layer | Examples | Role |
|---|---|---|
| Kernel | Linux | Manages hardware, processes, memory, filesystems |
| Core libraries | glibc | Provides the standard C library that programs link against |
| Shell / userland tools | bash, coreutils (`ls`, `cp`, `grep`, ...) | The command-line tools and interpreter users interact with |
| Init system | systemd, SysVinit | Boots the system and starts/manages background services |
| Package manager | APT, DNF/YUM, pacman | Installs, updates, and removes software |
| Desktop environment (optional) | GNOME, KDE Plasma, XFCE | Graphical user interface, not present on most servers |

A **distribution** is one particular combination and configuration of all of these layers, packaged together with an installer.


## 5. What Is a Linux Distribution?

Since the kernel and the GNU tools are both free to redistribute, many different groups and companies have combined them (plus their own installer, package manager, default configuration, and additional software) into a ready-to-use operating system. Each such bundle is called a **distribution** (often shortened to **distro**).

Distributions differ in:
- **Package format and manager** (`.deb`/APT, `.rpm`/DNF-YUM, pacman, etc.)
- **Release model** (fixed periodic releases vs. a continuously updated "rolling release")
- **Default init system, filesystem, and configuration conventions**
- **Target audience** (general desktop use, servers, security testing, embedded systems, etc.)
- **Support model** (community-supported vs. commercially supported with a paid subscription)


## 6. Major Distribution Families

Most distributions trace back to one of a small number of "parent" distributions, and inherit that parent's package format and conventions.

| Family | Package format | Notable members |
|---|---|---|
| Debian-based | `.deb` (APT) | Debian, Ubuntu, Linux Mint, Kali Linux, Raspberry Pi OS |
| Red Hat-based | `.rpm` (DNF/YUM) | Red Hat Enterprise Linux (RHEL), Fedora, Rocky Linux, AlmaLinux, CentOS Stream |
| SUSE-based | `.rpm` (Zypper) | openSUSE, SUSE Linux Enterprise |
| Arch-based | pacman | Arch Linux, Manjaro, EndeavourOS |
| Independent | varies | Slackware (one of the oldest, still actively maintained), Gentoo (source-based), Alpine Linux (musl/BusyBox-based, popular as a minimal container base image) |

Knowing a distribution's family is often enough to guess how software is installed and where configuration files live, even on a distro you've never used before.


## 7. Choosing a Distribution

There is no single "best" distribution - the right choice depends on the goal:

- **Learning / general desktop use** - Ubuntu, Fedora, and Linux Mint are popular starting points because of large communities and abundant documentation.
- **Production servers** - RHEL, Rocky Linux, AlmaLinux, Ubuntu LTS, or Debian Stable are common choices, valued for long support windows and release stability.
- **Certification study (e.g. LPIC-1, RHCSA)** - it's worth practicing on the family the exam targets, but the underlying concepts (permissions, processes, the shell) transfer across almost every distribution.
- **Rolling release, latest software** - Arch Linux or openSUSE Tumbleweed trade some stability for always having the newest package versions.
- **Minimal / container base images** - Alpine Linux is popular for its very small size.

**LTS** ("Long Term Support") releases receive security updates for an extended period (often 3-5+ years) and are generally preferred for production servers over short-lifecycle releases.


## Hands-on

Try these on your own Linux system (or a virtual machine):

```bash
cat /etc/os-release       # identify the distribution and version
uname -a                    # kernel name, version, and architecture
uname -r                     # kernel version only
lsb_release -a                 # distribution info (if lsb-release is installed)
hostnamectl                     # a systemd-based summary including OS and kernel info
```

Compare the output of `cat /etc/os-release` on two different distributions (or look up example output online for a distro you don't have access to) and identify which family each one belongs to.

```
Identify a system:
  cat /etc/os-release      distro name, version, ID
  uname -a                  kernel name/version/architecture
  hostnamectl                 systemd summary (OS + kernel)

Distribution families:
  Debian-based   -> .deb  -> APT      (Debian, Ubuntu, Mint, Kali)
  Red Hat-based  -> .rpm  -> DNF/YUM  (RHEL, Fedora, Rocky, Alma)
  SUSE-based     -> .rpm  -> Zypper   (openSUSE, SLE)
  Arch-based     -> pacman            (Arch, Manjaro)

Key history dates:
  1969-1970  Unix created (Bell Labs)
  1983        GNU Project launched
  1991         Linux kernel first released (Linus Torvalds)

LTS = Long Term Support: longer security-update window, preferred for servers.
```
