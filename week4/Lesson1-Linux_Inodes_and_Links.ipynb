## Inodes

Every filesystem object has an inode containing metadata such as:
- inode number
- UID
- GID
- permissions
- timestamps
- size
- data block pointers
- hard-link count

**Important:** filenames are NOT stored in the inode.



## Directory Entries

Directories map filenames to inode numbers.

```
report.txt ---> inode 1234
photo.jpg  ---> inode 5678
```

Therefore:
- one filename maps to one inode
- one inode may have many filenames (hard links)



## Pathname vs Filename

Example:

`/home/ali/Documents/report.txt`

- pathname = entire path
- filename = report.txt

Commands:

```bash
basename /home/ali/Documents/report.txt
dirname /home/ali/Documents/report.txt
```



## Viewing inode numbers

```bash
ls -i file
stat file
```

`stat` also shows the hard-link count.



## Hard Links

Create:

```bash
ln original.txt hard.txt
```

Both names reference the same inode.

Properties:

- same inode
- same contents
- editing one changes the other
- deleting one only removes one directory entry
- inode disappears only when link count becomes zero and no process keeps it open
- cannot normally link directories
- cannot cross filesystems



## Symbolic Links

Create:

```bash
ln -s target.txt shortcut
```

A symbolic link:

- has its own inode
- stores a pathname
- can cross filesystems
- can point to directories
- becomes broken if target disappears



## Hard Link vs Symbolic Link

|Hard|Symbolic|
|---|---|
|Same inode|Different inode|
|Cannot cross filesystems|Can cross filesystems|
|Cannot link directories|Can link directories|
|Cannot become broken by rename of inode|May become broken|
|Stores inode|Stores pathname|



## Applications

Hard links:
- save space
- backups
- snapshots

Symbolic links:
- software version management
- `/etc/alternatives`
- shared libraries
- Python (`python -> python3 -> python3.12`)
- moving directories transparently



## Useful Commands

```bash
ln file hardlink
ln -s target symlink
ls -li
stat file
readlink symlink
realpath symlink
find . -inum NUMBER
find -L .
```


````
echo "hello" > original.txt
ln original.txt hard.txt
ln -s original.txt sym.txt

ls -li original.txt hard.txt sym.txt
stat original.txt
stat hard.txt
stat sym.txt

echo "Linux" >> hard.txt
cat original.txt

readlink sym.txt
realpath sym.txt || true
````
