# Linux `tar` — Complete Educational Notebook

A comprehensive notebook on archiving with `tar`, including theory, options, labs, and exercises.

## 1. What is tar?

`tar` (**T**ape **Ar**chiver) is the standard Unix/Linux archiving utility. It combines files/directories into one archive while preserving metadata. By itself it **does not compress** data.

## 2. Archive vs Compression

Archiving combines files into one container. Compression reduces file size. `tar` archives; `gzip`, `bzip2`, and `xz` compress.

## 3. Why tar exists

Directories contain names, permissions, timestamps, symbolic links, and hierarchy. `tar` preserves all of these in one stream.

## 4. Archive format

.tar archives preserve directory trees, permissions, ownership (when possible), timestamps, hard links, symbolic links, and empty directories.

## 5. Basic syntax

General syntax:
```bash
tar [options] -f archive.tar files...
```

## 6. -c

Create a new archive.

## 7. -x

Extract an archive.

## 8. -t

List archive contents without extracting.

## 9. -f

The next argument is the archive filename.

## 10. -v

Verbose mode.

## 11. -C

Change to another directory before extraction.

## 12. --exclude

Exclude files or patterns from the archive.

## 13. Creating archives

Examples:
```bash
tar -cf backup.tar project/
tar -cvf backup.tar project/
```

## 14. Extracting

Examples:
```bash
tar -xf backup.tar
tar -xvf backup.tar
```

## 15. Listing

Examples:
```bash
tar -tf backup.tar
tar -tvf backup.tar
```

## 16. Extract one file

```bash
tar -xf backup.tar project/src/main.c
```

## 17. Extract elsewhere

```bash
mkdir restore
tar -xf backup.tar -C restore
```

## 18. gzip integration

`tar` doesn't compress automatically.
Create:
```bash
tar -czf backup.tar.gz project/
```
Extract:
```bash
tar -xzf backup.tar.gz
```

## 19. bzip2 integration

Create:
```bash
tar -cjf backup.tar.bz2 project/
```
Extract:
```bash
tar -xjf backup.tar.bz2
```

## 20. xz integration

Create:
```bash
tar -cJf backup.tar.xz project/
```
Extract:
```bash
tar -xJf backup.tar.xz
```

## 21. Appending

Append files (uncompressed tar only):
```bash
tar -rf backup.tar newfile.txt
```

## 22. Updating

Update only newer files:
```bash
tar -uf backup.tar project/
```

## 23. Concatenate

```bash
tar -Af archive1.tar archive2.tar
```

## 24. Delete entries

GNU tar supports:
```bash
tar --delete -f archive.tar file
```
(Not for compressed archives.)

## 25. Wildcards

Use `--wildcards` when extracting by pattern.

## 26. Preserving permissions

`tar` stores metadata. Restoring ownership usually requires root.

## 27. Incremental backups

GNU tar supports snapshot-based incremental archives using `--listed-incremental`.

## 28. Streaming

Examples:
```bash
tar -cf - project | gzip > backup.tar.gz
gzip -dc backup.tar.gz | tar -xf -
```

## 29. SSH backups

```bash
tar -cf - project | gzip | ssh host 'cat > backup.tar.gz'
```

## 30. Integrity

Use `gzip -t`, `bzip2 -t`, or `xz -t` after compression.

## 31. Common extensions

`.tar`, `.tar.gz`, `.tgz`, `.tar.bz2`, `.tar.xz`.

## 32. Common mistakes

Confusing archiving with compression; forgetting `-f`; using `j` instead of `J`; extracting into the wrong directory.

## 33. Performance

gzip is fastest; bzip2 often compresses better; xz usually provides the smallest archives but is slowest.

## 34. Real-world uses

Linux source releases, backups, Docker build contexts, software distribution, log archives.

## 35. Comparison table

|Format|Compression?|Typical tool|
|---|---|---|
|tar|No|tar|
|tar.gz|gzip|tar+gzip|
|tar.bz2|bzip2|tar+bzip2|
|tar.xz|xz|tar+xz|

## 36. Security

Inspect unknown archives with `tar -tf` before extraction. Extract into a temporary directory when possible.

## 37. Review

Practice all create/list/extract commands.

# Cheat Sheet

```bash
# Create
tar -cf archive.tar dir
tar -czf archive.tar.gz dir
tar -cjf archive.tar.bz2 dir
tar -cJf archive.tar.xz dir

# List
tar -tf archive.tar
tar -tzf archive.tar.gz
tar -tjf archive.tar.bz2
tar -tJf archive.tar.xz

# Extract
tar -xf archive.tar
tar -xzf archive.tar.gz
tar -xjf archive.tar.bz2
tar -xJf archive.tar.xz

# Extract elsewhere
tar -xzf archive.tar.gz -C restore

# Append (uncompressed tar only)
tar -rf archive.tar newfile

# Update (uncompressed tar only)
tar -uf archive.tar file

# Delete (GNU tar, uncompressed)
tar --delete -f archive.tar file
```
