# Linux Process Management — Complete Educational Notebook

Covers processes, PID, PPID, priorities, `ps`, process states, and `pstree`.

## Learning Objectives

Understand processes, multitasking, PID, PPID, process owners, priorities, `ps`, process states, and `pstree`.

## 1. What is a Process?

A **process** (or task) is a running instance of a program. Linux is a multitasking operating system, allowing many processes to execute concurrently.

## Program vs Process

A program is a passive executable file. A process is its active execution with memory, CPU state, open files, and environment.

## 2. PID

Each running process has a unique **Process ID (PID)** assigned by the kernel.

## 3. PPID

Every process has a **Parent Process ID (PPID)** indicating which process created it.

## 4. Process Owner

Each process runs on behalf of a user. The process owner is not necessarily the owner of every file it accesses.

## 5. Process Priority

Linux schedules processes using priorities. Nice values range from **-20** (highest priority) to **19** (lowest priority).

## Boot Sequence

Firmware → Boot Loader (GRUB) → Linux Kernel → PID 1 (`systemd` on most systems, historically `init`). The boot loader is not itself a Linux process.

## PID 1

PID 1 starts system services and adopts orphaned processes.

## 6. `ps`

`ps` displays process information.

## Common `ps` options

`ps`, `ps -e`, `ps -ef`, `ps aux`, `ps -u USER`, `ps -p PID`, `ps -l`, `ps -f`, `ps -eo pid,ppid,user,ni,pri,stat,cmd`.

## Understanding `ps` output

Common fields: UID, PID, PPID, NI, PRI, STAT, TTY, TIME, CMD.

## 9. States R and S

`R` = Running/runnable. `S` = Interruptible sleep waiting for an event.

## 10. States T and Z

`T` = Stopped (e.g., Ctrl+Z). `Z` = Zombie (terminated but not yet reaped by its parent).

## Other states

`D` = Uninterruptible sleep. `I` = Idle kernel thread (many systems).

## 11. `pstree`

`pstree` displays parent-child relationships as a process tree.
Examples:
```bash
pstree
pstree -p
pstree -u
```

## Useful Tools

`top`, `htop`, `pgrep`, `pidof`, `kill`, `nice`, `renice`, `jobs`, `bg`, `fg`.

## Review Questions

1. Difference between program and process?
2. What is PID?
3. What is PPID?
4. What is PID 1?
5. Difference between R and S?
6. Difference between T and Z?
7. What does `ps -ef` show?
8. What does `pstree -p` display?

## Answers

1. Program=disk file, process=running instance.
2. Unique process ID.
3. Parent process ID.
4. Usually `systemd`.
5. R=running, S=sleeping.
6. T=stopped, Z=zombie.
7. Full list of processes.
8. Process tree with PIDs.

````
echo "Current shell PID: $$"
echo "Parent PID: $PPID"

ps
echo
ps -ef | head
echo
ps -eo pid,ppid,user,ni,pri,stat,cmd | head

echo
pstree -p $$ 2>/dev/null || echo "pstree not installed"

sleep 30 &
PID=$!
echo "Sleep PID: $PID"
ps -p $PID -o pid,ppid,user,ni,pri,stat,cmd
kill $PID
wait $PID 2>/dev/null
````
