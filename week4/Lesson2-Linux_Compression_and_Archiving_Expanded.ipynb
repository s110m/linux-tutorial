## 1. Compression vs Archiving

Compression reduces the number of bytes needed to store data. Archiving combines multiple files/directories into one archive, usually preserving metadata and directory structure.

Examples:
- Archive only: `tar`
- Compress only: `gzip`, `bzip2`, `xz`
- Archive + Compress: `tar.gz`, `tar.bz2`, `tar.xz`

Files like **JPEG**, **MP3**, **AAC**, and **H.264** are already compressed formats. Compressing them again with gzip usually provides little or no benefit.

## 2. Lossless vs Lossy Compression

**Lossless** compression allows exact reconstruction of the original data.

Examples:
- gzip
- bzip2
- xz
- ZIP
- PNG
- FLAC

**Lossy** compression permanently discards some information.

Examples:
- JPEG
- MP3
- AAC
- H.264/H.265 video
- WebP (lossy mode)

Use lossless for programs, source code, databases, documents and backups. Use lossy for multimedia when a smaller size is acceptable.

## 3. Why Archiving is Needed

Compression programs generally compress one file. To compress an entire directory while preserving hierarchy, Linux commonly uses:

```bash
tar -cf archive.tar directory
gzip archive.tar
```

or

```bash
tar -czf archive.tar.gz directory
```


## 4. gzip

`gzip` is the most widely used general-purpose lossless compression utility on Unix/Linux.

Characteristics:
- Uses the DEFLATE algorithm.
- Fast compression/decompression.
- Good compatibility.
- Produces `.gz` files.
- Compresses one file at a time.


## 5. gzip Common Options

|Option|Meaning|
|---|---|
|`-d`|decompress|
|`-c`|write to stdout|
|`-k`|keep original file|
|`-r`|recursive|
|`-l`|list compressed file information|
|`-1`|fastest compression|
|`-9`|best compression|
|`-v`|verbose|
|`-f`|force overwrite|

Examples:

```bash
gzip file.txt
gzip -k file.txt
gzip -9 file.txt
gzip -d file.txt.gz
gunzip file.txt.gz
gzip -c file.txt > file.txt.gz
gzip -l file.txt.gz
```

## 6. bzip2

`bzip2` generally compresses better than gzip but is slower.

Characteristics:
- Burrows-Wheeler Transform
- Output extension `.bz2`
- Good for text and source code
- Slower than gzip

Useful options:

```bash
bzip2 file
bunzip2 file.bz2
bzip2 -k file
bzip2 -d file.bz2
```

## 7. xz

`xz` usually achieves the highest compression ratio among these three.

Characteristics:
- LZMA2 algorithm
- Extension `.xz`
- Excellent compression
- Slowest compression
- Popular for software distributions

Examples:

```bash
xz file
xz -k file
xz -d file.xz
unxz file.xz
xz -9 file
```

## 8. Comparison

|Program|Speed|Compression Ratio|Extension|
|---|---|---|---|
|gzip|Fast|Good|.gz|
|bzip2|Medium|Better|.bz2|
|xz|Slow|Best|.xz|

Typical choice:
- daily logs → gzip
- source archives → xz
- older compatibility → gzip


## 9. tar with Compression

Examples:

```bash
tar -cf archive.tar folder
tar -czf archive.tar.gz folder
tar -cjf archive.tar.bz2 folder
tar -cJf archive.tar.xz folder

tar -xzf archive.tar.gz
tar -xjf archive.tar.bz2
tar -xJf archive.tar.xz
```


## 10. Already Compressed Files

Usually these do **not** compress much further:

- jpg/jpeg
- png (already lossless compressed)
- gif
- mp3
- mp4
- avi with compressed codec
- pdf (often already compressed)
- zip
- rar
- 7z

Plain text, CSV, source code and log files often compress extremely well.
