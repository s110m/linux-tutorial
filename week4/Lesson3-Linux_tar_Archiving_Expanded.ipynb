# Linux tar: Archiving and Extraction

Expanded educational notebook.

## 1. What is tar?

`tar` (Tape ARchiver) is the standard Unix/Linux tool for **archiving** files and directories. By itself it **does not compress** data. It combines many files into one archive while preserving directory structure, permissions, timestamps, symbolic links, and (where possible) ownership.

## 2. Archiving vs Compression

`tar` creates archives. Compression is performed by tools such as **gzip**, **bzip2**, and **xz**. Modern `tar` can invoke these compressors automatically using command-line options.

## 3. tar -c

`-c` means **create** a new archive.

Example:
```bash
tar -c README.md
```
Normally `-c` is combined with `-f` to specify an output archive.

## 4. tar -x

`-x` means **extract** files from an archive.

Example:
```bash
tar -xf archive.tar
```

## 5. tar -f

`-f` tells tar that the next argument is the archive filename.

Examples:
```bash
tar -cf backup.tar project/
tar -xf backup.tar
```

## 6. tar -cf

Create an archive without compression.

```bash
tar -cf project.tar project/
```
List contents:
```bash
tar -tf project.tar
```

## 7. tar does not compress automatically

A plain `.tar` file is only an archive. To compress while archiving:

```bash
tar -czf archive.tar.gz folder/
```
Extract:
```bash
tar -xzf archive.tar.gz
```

## 8. tar -tf

List archive contents without extracting.

```bash
tar -tf archive.tar
tar -tzf archive.tar.gz
tar -tjf archive.tar.bz2
tar -tJf archive.tar.xz
```

## 9. tar with gzip

Create:
```bash
tar -czf backup.tar.gz project/
```
Extract:
```bash
tar -xzf backup.tar.gz
```
`z` tells tar to use **gzip**.

## 10. tar with bzip2

Create:
```bash
tar -cjf backup.tar.bz2 project/
```
Extract:
```bash
tar -xjf backup.tar.bz2
```
`j` selects **bzip2**.

## 11. tar with xz

Create:
```bash
tar -cJf backup.tar.xz project/
```
Extract:
```bash
tar -xJf backup.tar.xz
```
Capital `J` selects **xz**.

## 12. Useful options

|Option|Meaning|
|---|---|
|`-v`|verbose|
|`-C dir`|change directory before extract|
|`--exclude=PATTERN`|exclude files|
|`-t`|list contents|
|`-f`|archive filename|

## 13. Practical workflow

1. Archive a project.
2. Verify contents using `tar -tf`.
3. Extract into another directory using `-C`.
4. Compare extracted files.

## 14. Common mistakes

- Thinking `.tar` is compressed.
- Forgetting `-f`.
- Using lowercase `j` instead of uppercase `J` for xz.
- Extracting into the wrong directory.
