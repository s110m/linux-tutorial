# Linux Boot Process, Firmware, BIOS/UEFI, GRUB, and Recovery
This notebook is an educational guide with explanations, examples, and review questions.

## 1. Firmware

### What is firmware?
Firmware is low-level software stored in non-volatile memory on a device. It starts the hardware before any operating system exists.

**Analogy:** Firmware is like the building manager who unlocks the building every morning before employees (the OS) can start working.

**Example:** UEFI firmware on a PC, firmware inside an SSD, Wi-Fi router firmware.

**Why do we need it?**
- Initializes CPU, RAM, keyboard, disks, etc.
- Performs hardware checks.
- Finds a bootable device.
- Starts a boot loader.

**Firmware vs ROM**
- Firmware = the software.
- ROM = a storage technology where firmware was traditionally stored.
- Modern firmware is usually stored in rewritable flash memory, not true ROM.


## 2. Boot and Boot Loader

### Boot
Booting is the process of starting a computer until the operating system is running.

### Boot loader
A boot loader is a small program started by firmware. Its job is to locate and load an operating system kernel into RAM and transfer execution to it.


## 3. MBR

### Master Boot Record (MBR)
MBR is the first 512-byte sector of an MBR-partitioned disk.

It contains:
- Boot code
- Partition table
- Boot signature

**Firmware (BIOS)** reads the MBR from the selected boot disk and executes its boot code.


## 4. Firmware → Boot Loader → OS

The normal BIOS boot chain is:

Firmware (BIOS) → Boot Loader (GRUB) → Linux Kernel → init/systemd → User space

With UEFI:

UEFI Firmware → EFI boot manager → GRUB EFI program → Linux Kernel


## 5. GRUB

### GRUB
GRUB (GRand Unified Bootloader) is a boot loader.

Features:
- Boots Linux
- Boots Windows
- Boots BSD
- Menu interface
- Recovery options


## 6. Boot loader responsibilities

A boot loader must:
- Understand disk partitions
- Read supported filesystems (e.g. ext2/ext3/ext4, FAT for EFI)
- Locate the Linux kernel and initramfs
- Load them into RAM
- Pass kernel parameters
- Jump to the kernel


## 7. GRUB stages

### Classical BIOS GRUB
Historically:
- Stage 1: stored in MBR, very small.
- Stage 1.5 (optional): stored after the MBR, understands filesystems.
- Stage 2: full GRUB from /boot/grub.

Modern GRUB is more modular, but the stage explanation is still useful educationally.


## 8. Multiple operating systems

GRUB can present a menu allowing Linux, Windows, and other installed operating systems to boot.

## 9. Dual boot

Recommended:
1. Install Windows first.
2. Shrink Windows partition.
3. Install Linux into free space.

Reason:
Windows installers often overwrite the boot configuration (especially on BIOS systems and sometimes EFI boot order), so installing Linux afterward allows GRUB to detect Windows automatically.

On UEFI systems Windows does **not always delete GRUB**, but it may change the default boot entry.


## 10. Removing Linux

If Linux is removed carelessly:
- GRUB files may disappear.
- Firmware may still try to start GRUB.
- Windows itself is usually still intact.

Fix:
- Restore Windows Boot Manager or
- Reinstall GRUB if Linux remains.

Do **not** simply delete Linux partitions first without restoring a working boot loader.


## 11. BIOS vs UEFI

UEFI is the modern replacement for BIOS.

Advantages:
- Larger disks (GPT)
- Faster startup
- Secure Boot
- Better graphical interface
- Better extensibility

EFI = Extensible Firmware Interface.
UEFI = Unified EFI.


## 12. Where is the kernel?

Linux kernels are usually stored in **/boot**, for example:

/boot/vmlinuz-6.x
/boot/initrd.img-6.x

On EFI systems, the EFI System Partition is usually mounted at **/boot/efi**.


## 13. Booting manually from GRUB

You can interrupt GRUB (often by pressing **Esc** or **Shift** on BIOS systems).

Typical commands:

```
ls
set root=(hd0,gpt2)
linux /boot/vmlinuz-... root=/dev/sda2 ro
initrd /boot/initrd.img-...
boot
```

Exact paths depend on your system.


## 14. Repair using USB

Yes.

Boot a Linux live USB.

Possible repairs:
- Mount Linux partitions.
- Chroot into installed Linux.
- Reinstall GRUB.
- Repair configuration.

Usually you **do not need to reinstall Linux**, and your personal files can often be preserved.

Example:

```
sudo mount /dev/sda2 /mnt
sudo mount /dev/sda1 /mnt/boot/efi   # UEFI
sudo grub-install --boot-directory=/mnt/boot ...
sudo update-grub
```

Or use tools like Boot-Repair.


## 15. GRUB boot command

`boot`

This transfers execution to the kernel that has already been loaded into memory.
It does **not** search for the kernel.


## 16. Before boot

Before running `boot`, GRUB typically needs:

1. Set the correct root partition.
2. Load the kernel (`linux` command).
3. Load initramfs (`initrd` command).
4. Optionally set kernel parameters.
5. Execute `boot`.
