# Linux Vim Editor

Educational notebook introducing the Vim command-line editor.

## 1. What is Vim?

Vim (**Vi IMproved**) is a powerful **command-line text editor** available on almost every Linux/Unix system. It is an improved version of the original `vi`.

## 2. Why do we need command-line editors?

Graphical editors are not always available. Examples:

- Working on a remote server over SSH.
- Recovering a system in rescue mode.
- Minimal Linux installations.
- Editing configuration files before the graphical interface starts.

In these situations, editors like **vim** or **nano** are essential.

## 3. Nano vs Vim

`nano` is beginner-friendly because it shows shortcuts on the screen.

`vim` has a steeper learning curve but is preferred by many professionals because it is:

- extremely fast,
- available almost everywhere,
- keyboard-driven,
- highly configurable,
- powerful for programming and system administration.

## 4. Vim modes

The two most important modes are:

- **Command (Normal) mode**: navigate and execute commands.
- **Insert mode**: type text.

Press **Esc** to return to Command mode.

Enter Insert mode using:

```text
i   insert before cursor
I   insert at line start
a   append after cursor
A   append at line end
o   open new line below
O   open new line above
```

## 5. Save and Quit

Always return to Command mode with **Esc** first.

```text
:w      save
:q      quit
:q!     quit without saving
:wq     save and quit
:x      save if modified and quit
ZZ      save (if needed) and quit
:w filename   save as another file
```

## 6. Movement

```text
h  left
j  down
k  up
l  right
0  beginning of line
$  end of line
w  next word
b  previous word
G  last line
:n go to line n
Ctrl+f page forward
Ctrl+b page backward
```

## 7. Delete

```text
x   delete character under cursor
X   delete character before cursor
dd  delete current line
ndd delete n lines
```

## 8. Undo and Redo

```text
u      undo
Ctrl+r redo
```
(Some references write `:redo`, but the common shortcut is **Ctrl+r**.)

## 9. Copy and Paste

```text
yy    copy current line
nyy   copy n lines
p     paste after cursor
P     paste before cursor
```

## 10. Visual Mode

Press:

```text
v
```

for character-wise visual mode.

Other useful modes:

```text
V        line-wise selection
Ctrl+v   block (column) selection
```

After selecting text:

```text
y   copy
d   cut/delete
```

## 11. Search

```text
/pattern
```
Search forward.

```text
?pattern
```
Search backward.

```text
n
```
Next match.

```text
N
```
Previous match.

## 12. Common mistakes

- Forgetting to press **Esc** before commands.
- Typing `:wq` while still in Insert mode.
- Confusing `x` with `X`.
- Thinking Vim is broken when actually in the wrong mode.

## Answers

1. Works everywhere and is powerful.
2. Command executes commands; Insert edits text.
3. `i`, `a`, `o`, etc.
4. `:wq`, `:x`, or `ZZ`.
5. `x` deletes under cursor; `X` deletes before cursor.
6. `p` pastes after; `P` before.
7. `u` and `Ctrl+r`.
8. A mode for selecting text.
