# Linux `.deb` Packages and `dpkg`

Concise educational notebook.

## 1. `.deb` packages

A **`.deb`** file is the standard package format used by Debian-based systems such as Debian, Ubuntu, and Linux Mint. It contains software, metadata, documentation, configuration files, and installation scripts.

## 2. What is `dpkg`?

`dpkg` is the **low-level package management tool** for Debian-based systems. It installs, removes, and queries packages, but **does not automatically resolve dependencies**.

## 3. Package database

Package managers keep a database of installed packages and files. This allows future tasks such as upgrades, verification, removal, and determining which package owns a file.

## 4. Install a package

```bash
sudo dpkg -i package.deb
```
If dependencies are missing, use:
```bash
sudo apt install -f
```

## 5. Remove vs Purge

| Command | Program removed | Config files removed |
|---|:---:|:---:|
|`dpkg -r package`|✅|❌|
|`dpkg -P package`|✅|✅|

`-P` means **purge**.

## 6. Useful `dpkg` options

```bash
dpkg -i file.deb
dpkg -r package
dpkg -P package
dpkg -l
dpkg -L package
dpkg -c file.deb
dpkg -I file.deb
dpkg -S /bin/cat
dpkg -s package
```

## 7. `dpkg -c`

Lists the files contained inside a `.deb` package **without installing it**.

## 8. `dpkg -S /bin/cat`

Searches the package database to determine which installed package owns a file.

Typical output:
```text
coreutils: /bin/cat
```

## 9. `dpkg` vs `apt`

`dpkg` is low-level and does not resolve dependencies automatically. `apt` is the recommended high-level tool for installing packages from repositories.

## 10. Common mistakes

- Confusing `.deb` with `dpkg`.
- Thinking `dpkg` downloads dependencies.
- Forgetting `dpkg -r` keeps configuration files.
