# Linux APT Package Manager

Educational notebook.

## What is APT?

APT (Advanced Package Tool) is the high-level package manager for Debian-based systems. It downloads packages from repositories, resolves dependencies automatically, upgrades software, and uses `dpkg` underneath.

## APT vs dpkg

`dpkg` is the low-level installer and package database. `apt` resolves dependencies, downloads packages, and then calls `dpkg` to install/remove them.

## Repositories

APT is connected to software repositories. Repository definitions are commonly stored in `/etc/apt/sources.list` and `/etc/apt/sources.list.d/`. You can add repositories (e.g. with `add-apt-repository`) and then run `sudo apt update`.

## Dependency resolution

APT checks package versions, dependencies, conflicts, and repository metadata before installation.

## `apt` vs `apt-get`

`apt` is recommended for interactive use. `apt-get` is older and often preferred in scripts because of its stable interface.

## `apt update` vs `apt upgrade`

`apt update` downloads package lists. `apt upgrade` installs newer versions of installed packages using those updated lists.

## Package cache

Downloaded `.deb` files are cached in:
```bash
ls -l /var/cache/apt/archives
```

## Upgrade one package

Upgrade only one package:
```bash
sudo apt install package_name
```

## Search packages

```bash
apt search python
apt show vim
apt list --installed
apt list --upgradable
```

## Useful commands

```bash
sudo apt update
sudo apt upgrade
sudo apt full-upgrade
sudo apt install package
sudo apt remove package
sudo apt purge package
sudo apt autoremove
sudo apt clean
sudo apt autoclean
```

# Cheat Sheet

```bash
apt update
apt upgrade
apt install package
apt remove package
apt purge package
apt autoremove
apt search keyword
apt show package
apt list --upgradable
```

- APT = high-level manager
- dpkg = low-level installer/database
- update = refresh package lists
- upgrade = install updates
