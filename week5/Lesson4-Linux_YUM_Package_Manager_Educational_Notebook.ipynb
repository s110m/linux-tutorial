# Linux YUM Package Manager

Educational notebook about **YUM**, RPM packages, repositories, dependencies, and common commands.

## 1. What is YUM?

YUM (**Yellowdog Updater, Modified**) is a **high-level package manager** for RPM-based Linux distributions. It downloads packages from repositories, resolves dependencies automatically, and installs them using the low-level `rpm` tool.

## 2. YUM vs RPM

`rpm` is the low-level package tool. `yum` downloads packages, resolves dependencies, updates repositories, and then calls `rpm` to perform installation or removal.

## 3. Repositories

YUM installs software from repositories. Repository configuration files are usually stored in:

```text
/etc/yum.repos.d/
```

Each `.repo` file defines one or more repositories.

## 4. Dependency Resolution

Before installing a package, YUM checks:
- required dependencies
- package conflicts
- installed versions
- repository metadata

It automatically downloads the required packages.

## 5. YUM vs DNF

Modern Fedora and newer RHEL-family systems use **DNF**, which replaced YUM as the default package manager. Most YUM commands still work because `yum` is often a compatibility wrapper around DNF.

## 6. Update Repository Metadata

```bash
sudo yum check-update
```
Checks for package updates and refreshes metadata as needed.

## 7. Install Packages

```bash
sudo yum install vim
sudo yum install git
```

## 8. Remove Packages

```bash
sudo yum remove vim
```

## 9. Upgrade Packages

Upgrade all packages:
```bash
sudo yum update
```
Upgrade a single package:
```bash
sudo yum update vim
```

## 10. Search Packages

```bash
yum search python
yum info vim
yum list installed
yum list updates
```

## 11. Clean Cache

```bash
sudo yum clean all
sudo yum makecache
```
Cleaning removes cached metadata/packages. `makecache` rebuilds repository metadata.

## 12. Installed Package Database

Although YUM is used, the installed packages are still tracked by the RPM database because YUM calls `rpm` underneath.

## 13. Useful Commands

```bash
yum repolist
yum provides /bin/ls
yum history
yum history undo ID
```

## 14. Other Package Managers

## 15. Common Mistakes

- Confusing YUM with RPM.
- Thinking YUM compiles software.
- Forgetting to refresh metadata.
- Editing `.repo` files incorrectly.
