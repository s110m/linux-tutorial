# Linux Packages, RPM, DEB, and Package Managers

A concise educational notebook about Linux software packaging.

## What is a package?

A **package** is a bundle containing software, metadata, configuration files, documentation, and installation information.

## What is a packager?

A **packager** prepares software into installable formats such as RPM or DEB.

## What is a package manager?

A **package manager** installs, upgrades, removes, verifies, and tracks software packages.

Examples: RPM, DNF, YUM, APT, dpkg, Pacman, Zypper.

## Why not always compile from source?

Manual compilation requires build tools, manual dependency handling, harder upgrades/uninstalls, and bypasses the package database. Package managers automate these tasks.

## RPM

**RPM** originally meant **Red Hat Package Manager**. It is both a package format and a low-level package-management tool.

## Dependencies

A dependency is another package required by a program. Modern tools like DNF and APT resolve dependencies automatically.

## Useful RPM commands

```bash
rpm -qa
rpm -qi bash
rpm -ql bash
rpm -qf /bin/bash
rpm -ivh file.rpm
rpm -Uvh file.rpm
rpm -e package
rpm -V package
```

## Running without installing

Some programs can run directly (portable applications or `./program`). Installing usually copies files into standard locations and registers them in the package database.

## RPM vs DEB

`.rpm` and `.deb` are package formats.

RPM-based: Fedora, RHEL, Rocky, AlmaLinux.

DEB-based: Debian, Ubuntu, Linux Mint.

## RPM vs DNF

`rpm` works directly with RPM files but does not resolve dependencies automatically. `dnf` manages RPM packages and resolves dependencies.

## DEB tools

`dpkg` is the low-level DEB tool. `apt` is the recommended high-level package manager because it resolves dependencies.

## Common mistakes

Do not confuse `.rpm` with `rpm`; package formats with package managers; or installation with compilation.

# Cheat Sheet

```bash

rpm -ql package

rpm -e package
rpm -V package
```

- Package = software bundle
- Packager = creates packages
- Package manager = installs/manages packages
- Dependency = required package
- `.rpm` / `.deb` = formats
